```python
folder = "New folder"

files = [
    "customers.csv",
    "products.csv",
    "sellers.csv",
    "locations.csv",
    "orders.csv",
    "order_items.csv",
    "payments.csv",
    "shipping.csv",
    "returns.csv"
]

data = {}

for file in files:
    data[file.replace(".csv", "")] = pd.read_csv(
        os.path.join(folder, file)
    )

print("All files loaded successfully!")
```

```python
for name, df in data.items():
    print(f"{name:15} → {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
for name, df in data.items():
    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)
    print(df.columns.tolist())
```

```python
for name, df in data.items():
    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)
    display(df.head(3))
```

```python

for name, df in data.items():
    print(f"\n{'='*50}")
    print(f"Dataset: {name}")
    print("="*50)
    print(df.dtypes)
```

```python
for name, df in data.items():
    print(f"\n{name.upper()}")
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values found.")
    else:
        print(missing.sort_values(ascending=False))
```

```python
for name, df in data.items():
    print(f"{name:15} → {df.duplicated().sum():,} duplicate rows")
```

```python
for name in ["orders", "payments"]:
    df = data[name]

    print(f"\n{'='*50}")
    print(f"Dataset: {name.upper()}")
    print("="*50)

    # Show duplicate records, including the original rows
    duplicates = df[df.duplicated(keep=False)]

    print("Rows involved in exact duplicates:", len(duplicates))
    display(duplicates.head(10))

    # Check whether the primary ID is duplicated
    id_col = "order_id" if name == "orders" else "payment_id"

    print(f"Duplicate {id_col} values:", df[id_col].duplicated().sum())
```

```python

numeric_columns = {
    "products": ["list_price", "cost_price", "weight_kg"],
    "orders": ["order_total"],
    "order_items": ["quantity", "unit_price", "discount_amount"],
    "payments": ["amount"],
    "returns": ["refund_amount"]
}

for name, columns in numeric_columns.items():
    print(f"\n{'='*40}")
    print(f"Dataset: {name.upper()}")

    for col in columns:
        df = data[name]
        converted = pd.to_numeric(df[col], errors="coerce")

        invalid = df[col].notna() & converted.isna()

        print(f"\n{col}")
        print("Invalid numeric values:", invalid.sum())

        if invalid.any():
            print("Examples:", df.loc[invalid, col].unique()[:10])
```

```python

for name, df in data.items():
    missing_pct = (df.isnull().mean() * 100).round(2)
    missing_pct = missing_pct[missing_pct > 0]
    missing_pct = missing_pct.sort_values(ascending=False)

    print(f"\n{name.upper()} — Missing Values (%)")
    display(missing_pct.to_frame(name="Missing %"))
```
